Basic Validation

```python
print("========== BASIC VALIDATION ==========")

print(f"Users:              {len(users):,}")
print(f"Devices:            {len(devices):,}")
print(f"Applications:       {len(applications):,}")
print(f"Authentication:     {len(auth_events):,}")
print(f"Access:             {len(access_events):,}")
print(f"Privilege:          {len(privilege_events):,}")
```

```python
assert len({u["user_id"] for u in users}) == len(users)

assert len({d["device_id"] for d in devices}) == len(devices)

assert len({a["application_id"] for a in applications}) == len(applications)

assert len({e["event_id"] for e in auth_events}) == len(auth_events)

assert len({e["event_id"] for e in access_events}) == len(access_events)

assert len({e["event_id"] for e in privilege_events}) == len(privilege_events)

print("✓ ID uniqueness validation passed")
```

```python
user_ids = {u["user_id"] for u in users}
device_ids = {d["device_id"] for d in devices}

assert all(
    device["user_id"] in user_ids
    for device in devices
)

print("✓ Device → User referential integrity passed")
```

```python
assert all(
    event["user_id"] in user_ids
    for event in auth_events
)

assert all(
    event["device_id"] in device_ids
    for event in auth_events
)

print("✓ Authentication referential integrity passed")
```

```python
assert all(
    event["user_id"] in user_ids
    for event in access_events
)

assert all(
    event["device_id"] in device_ids
    for event in access_events
)

print("✓ Access referential integrity passed")
```

```python
device_to_user = {
    device["device_id"]: device["user_id"]
    for device in devices
}

assert all(
    device_to_user[event["device_id"]] == event["user_id"]
    for event in auth_events
)

assert all(
    device_to_user[event["device_id"]] == event["user_id"]
    for event in access_events
)

print("✓ User → Device consistency passed")
```

```python
invalid_auth_events = [
    event
    for event in auth_events
    if device_to_user.get(event["device_id"]) != event["user_id"]
]

invalid_access_events = [
    event
    for event in access_events
    if device_to_user.get(event["device_id"]) != event["user_id"]
]

print(f"Invalid authentication events: {len(invalid_auth_events):,}")
print(f"Invalid access events:         {len(invalid_access_events):,}")
```

```python
users_without_devices = [
    user["user_id"]
    for user in users
    if not simulator.user_devices.get(user["user_id"])
]

print(f"Users without devices: {len(users_without_devices)}")
```

```python
import json
import os

source_path = "/Volumes/workspace/cyberaccess/source_data"

os.makedirs(source_path, exist_ok=True)


def write_json_lines(events, file_path):
    with open(file_path, "w") as f:
        for event in events:
            f.write(json.dumps(event, default=str) + "\n")


write_json_lines(
    auth_events,
    f"{source_path}/authentication_events.json"
)

write_json_lines(
    access_events,
    f"{source_path}/access_events.json"
)

write_json_lines(
    privilege_events,
    f"{source_path}/privilege_events.json"
)
```
